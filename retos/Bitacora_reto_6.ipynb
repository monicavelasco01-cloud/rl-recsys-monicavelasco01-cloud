## Conclusiones

Los resultados muestran que el recomendador basado en factorización implícita logró superar el desempeño de la línea base “repetir lo propio” en las condiciones exigidas por el reto. Esto responde a la primera pregunta: “¿Por qué gana repetir lo propio en estos datos?” La razón es que el historial previo de cada estudiante contiene una señal fuerte sobre sus interacciones futuras; es decir, en estos datos existe una tendencia importante a que los estudiantes vuelvan a interactuar con elementos que ya habían practicado. Sin embargo, el resultado también muestra que no basta con mirar solamente el Recall: la factorización obtuvo un nDCG@10 de 0,5634, menor que el 0,5880 de “repetir lo propio”. Por tanto, la factorización encuentra ligeramente más elementos relevantes, pero los coloca en posiciones menos favorables dentro de las primeras 10 recomendaciones.

Respecto a la segunda pregunta, “¿qué dice esto sobre el dominio?”, los resultados sugieren que en ASSISTments el comportamiento de los estudiantes presenta una fuerte relación con su propio historial, por lo que recomendar exclusivamente elementos nuevos no resulta adecuado para este conjunto de datos. La explicación de la diferencia entre la predicción y el resultado real debe centrarse en que la factorización aprovecha patrones de interacción que van más allá de la simple repetición, permitiéndole alcanzar 0,7142 de Recall, ligeramente por encima de 0,7009, aunque sin superar a la línea base en nDCG. Además, el experimento confirma lo visto en el Taller A: el mismo recomendador, sin cambiar una sola línea de código, pasó de 0,5634 de nDCG con partición temporal a 0,8395 con partición aleatoria, un aumento aproximado del 49 %.
