Ante esto, nuestra predicción fue que al implementar una política $\epsilon$-ávida con un valor de 0.2, romperíamos este bucle forzando la exploración, lo que nos permitiría superar el umbral de +0.1400 de ganancia y alcanzar una cobertura de al menos 15 habilidades. La ejecución confirmó nuestra hipótesis: el agente pasó de estar "clavado" a tocar 19.9 brazos, logrando una ganancia de +0.1445, lo cual demuestra que la exploración activa es indispensable para evitar el estancamiento prematuro en entornos de aprendizaje.

Lejos de ser un resultado negativo, esta es nuestra lección fundamental: en entornos no estacionarios donde las habilidades se agotan, la diversidad es el factor determinante. Hemos aprendido que, antes de añadir sofisticación, debemos medir si la estrategia de distribución básica ya está resolviendo el problema, lo que nos permite evitar la complejidad innecesaria en nuestros sistemas de recomendación.
